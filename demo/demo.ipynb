```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
np.random.seed(42)
```

```python
fs = 100
duration = 20

t = np.arange(0, duration, 1/fs)

vibration = 0.2 * np.sin(2 * np.pi * 10 * t)
vibration += 0.05 * np.random.randn(len(t))
```

```python
fault_start = 12

fault_region = t >= fault_start

vibration[fault_region] += (
    0.6 * np.sin(2 * np.pi * 30 * t[fault_region])
)
```

```python
plt.figure(figsize=(12, 4))
plt.plot(t, vibration)
plt.xlabel("Time (seconds)")
plt.ylabel("Acceleration")
plt.title("Raw Vibration Signal")
plt.show()
```

```python
temp_time = np.arange(0, duration, 1)

temperature = 45 + 0.2 * np.random.randn(len(temp_time))

temperature[temp_time >= fault_start] += np.linspace(
    0, 20, np.sum(temp_time >= fault_start)
)
```

```python
plt.figure(figsize=(12, 4))
plt.plot(temp_time, temperature, marker="o")
plt.xlabel("Time (seconds)")
plt.ylabel("Temperature °C")
plt.title("Temperature Sensor")
plt.show()
```

```python
current_fs = 10

current_time = np.arange(0, duration, 1/current_fs)

current = 2.2 + 0.1 * np.random.randn(len(current_time))

current[current_time >= fault_start] += 1.5
```

```python
plt.figure(figsize=(12, 4))
plt.plot(current_time, current)
plt.xlabel("Time (seconds)")
plt.ylabel("Current A")
plt.title("Motor Current")
plt.show()
```
